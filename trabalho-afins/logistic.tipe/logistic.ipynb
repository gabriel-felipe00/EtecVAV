```python
import numpy as np
from sklearn.linear_model import LogisticRegression

#dados simulação vibração do motor (x) a falha (y) a = não, 1 = sim)

vibracao = np.array([[1.2], [2.5],[3.1],[4.8],[5.5]])
falha= np.array([0,0,0,1,1])

#criando = treinando o modelo

modelo = LogisticRegression()

modelo.fit(vibracao, falha)
resposta =modelo.predict_proba([[4.0]])[0][1]
print(F"pega essa cattinga: {resposta:.2f}")
```
