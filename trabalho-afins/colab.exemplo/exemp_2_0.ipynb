```python
#impotando bibliotecas necessarias

import numpy as np
import matplotlib.pyplot as plit

from sklearn.cluster import KMeans

# 1. criação dos dados simulados (contextualizados

#------------------------------------

#cenario: temos sensores em diferentes pontos de planta da industrial
# medindo temperatura (°C) e umidade relativa do ar (%).
#queremos descobrir, sem rotulos, quais sensores estão em condições
#semelhanças.

sensores = np.array([
 [22,65],[23,63],[24,66],[28,80],[27,78],
[30,85],[21,60],[20,58],[29,83],[25,70]
])

#------------------------------------

# 2. aplicação de algoritmo K-means

#definições que queremos formar 2 grupos (clusters)
kmeans = KMeans(n_clusters=2, random_state=0, n_init='auto')

kmeans.fit(sensores)

#----------------------------------

# 3. visaulização dos algoritmos

#cada ponto sera colorido conforme o grupo ao qual foi atributado

plit.scatter (
 sensores[:,0],
sensores [:,1],
c=kmeans.labels_,
cmap= 'viridis'
)

plit.scatter(
  kmeans.cluster_centers_[:,0],
  kmeans.cluster_centers_[:,1],
  s=200,
  c= 'red',
marker= 'X',
label='Centros dos grupos'
)

plit.xlabel("temperatura (°C)")
plit.ylabel("umidade (%)")
plit.title ("agrupamento de sensores com K-Means")
plit.legend()
plit.grid(True)

plit.show()
```
